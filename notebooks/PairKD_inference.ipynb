# PairKD: original PKP2 inference

Run from a checkout of this repository. In Colab, mount Drive yourself and set REPOSITORY and MODEL_ZIP to your copies. This notebook does not fetch unpublished URLs or run hidden/public evaluation. Runtime dependencies must already be installed as documented in docs/ENVIRONMENTS.md. New notebook/CLI GPU acceptance is pending.

```python
from pathlib import Path
import subprocess, sys
REPOSITORY = Path.cwd()  # change to the repository checkout
MODEL_ZIP = Path('/path/to/PairKD-PKP2-seed13-epoch4-v0.1.0.zip')
ASSETS_ROOT = REPOSITORY / 'my_assets'
INPUT = REPOSITORY / 'examples/functions.jsonl'
OUTPUT = REPOSITORY / 'predictions.jsonl'
RUN_REAL_INFERENCE = False
assert (REPOSITORY / 'cli.py').is_file()
```

```python
subprocess.run([sys.executable, str(REPOSITORY/'scripts/verify_repository.py')], check=True)
subprocess.run([sys.executable, str(REPOSITORY/'cli.py'), 'infer', '--input', str(INPUT), '--output', str(OUTPUT), '--assets-root', str(ASSETS_ROOT), '--dry-run'], check=True)
```

```python
if RUN_REAL_INFERENCE:
    if not ASSETS_ROOT.exists():
        subprocess.run([sys.executable, str(REPOSITORY/'scripts/import_model_zip.py'), '--model-zip', str(MODEL_ZIP), '--assets-root', str(ASSETS_ROOT)], check=True)
    subprocess.run([sys.executable, str(REPOSITORY/'cli.py'), 'verify-assets', '--assets-root', str(ASSETS_ROOT)], check=True)
    subprocess.run([sys.executable, str(REPOSITORY/'cli.py'), 'infer', '--input', str(INPUT), '--output', str(OUTPUT), '--assets-root', str(ASSETS_ROOT), '--device', 'cuda:0'], check=True)
else:
    print('Real inference disabled; zero model calls.')
```
